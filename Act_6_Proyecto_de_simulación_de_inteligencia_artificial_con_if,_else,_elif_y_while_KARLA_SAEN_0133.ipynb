```python
# --- SIMULADOR DE ROBO-ADVISOR FINANCIERO: PORTAFOLIOS DE INVERSIÓN ---

print("--- Cargando Cerebro Lógico de Robo-Advisor IA ---")
print("Este sistema automatizado estructurará su portafolio ideal de inversión.")

# Variable de control para el ciclo de interactividad
ejecutar_sistema = "si"

# Estructura repetitiva principal (while) para mantener la IA activa
while ejecutar_sistema == "si" or ejecutar_sistema == "s":
    print("\n[NUEVA EVALUACIÓN DE PERFIL DE INVERSIÓN]")

    # Entrada de datos (input) y conversión de tipos
    nombre = input("Ingrese su nombre: ")
    capital = float(input("¿Cuánto capital desea colocar bajo gestión? (USD): "))
    riesgo = int(input("Nivel de tolerancia al riesgo (1 al 10): "))
    plazo_anios = int(input("¿A cuántos años planea retirar su inversión?: "))

    print(f"\nIA Robo-Advisor: Procesando métricas y algoritmos para {nombre}...")

    # Lógica de Decisión de la IA utilizando condicionales y operadores relacionales/lógicos

    # Caso 1: Perfil Conservador (Bajo Riesgo) -> Enfocado en proteger capital
    if riesgo <= 3 and capital > 0:
        tipo_portafolio = "Portafolio Conservador (80% Renta Fija, 20% Acciones Defensivas)"
        rendimiento_anual = 0.05  # 5% de retorno anual estimado

        # Operadores aritméticos para proyección básica de interés compuesto simple
        ganancia_estimada = capital * rendimiento_anual * plazo_anios
        total_final = capital + ganancia_estimada

        print(f"IA Robo-Advisor: Perfil Detectado: INVERSOR CONSERVADOR.")
        print(f"IA Robo-Advisor: Asignación Recomendada: {tipo_portafolio}.")
        print(f"IA Robo-Advisor: Retorno proyectado en {plazo_anios} años: ${ganancia_estimada:.2f}.")
        print(f"IA Robo-Advisor: Saldo final estimado en cuenta: ${total_final:.2f}.")

    # Caso 2: Perfil Moderado con capital significativo -> Balance entre crecimiento y seguridad
    elif riesgo > 3 and riesgo <= 7 and capital >= 5000:
        tipo_portafolio = "Portafolio Moderado Equilibrado (50% Fondos Indexados, 50% Bonos)"
        rendimiento_anual = 0.08  # 8% de retorno anual estimado

        ganancia_estimada = capital * rendimiento_anual * plazo_anios
        total_final = capital + ganancia_estimada

        print(f"IA Robo-Advisor: Perfil Detectado: INVERSOR MODERADO.")
        print(f"IA Robo-Advisor: Asignación Recomendada: {tipo_portafolio}.")
        print(f"IA Robo-Advisor: Este portafolio busca el crecimiento mitigando caídas severas.")
        print(f"IA Robo-Advisor: Saldo final estimado en cuenta: ${total_final:.2f}.")

    # Caso 3: Perfil Agresivo -> Maximización de rendimientos en mercados volátiles
    elif riesgo > 7 and riesgo <= 10:
        tipo_portafolio = "Portafolio Agresivo de Crecimiento (90% Acciones Tecnológicas, 10% Criptoactivos)"
        rendimiento_anual = 0.14  # 14% de retorno anual estimado

        ganancia_estimada = capital * rendimiento_anual * plazo_anios
        total_final = capital + ganancia_estimada

        print(f"IA Robo-Advisor: Perfil Detectado: INVERSOR AGRESIVO.")
        print(f"IA Robo-Advisor: Asignación Recomendada: {tipo_portafolio}.")
        print(f"IA Robo-Advisor: Retorno proyectado en {plazo_anios} años: ${ganancia_estimada:.2f}.")
        print(f"IA Robo-Advisor: Saldo final estimado en cuenta: ${total_final:.2f}.")

    # Caso 4: Manejo de Capital Insuficiente para Perfil Moderado, pero no Conservador.
    elif riesgo > 3 and riesgo <= 7 and capital < 5000:
        print("\nIA Robo-Advisor: Con el capital actual, un perfil moderado no es óptimo.")
        print("IA Robo-Advisor: Recomendamos iniciar con un Perfil Conservador para proteger su capital o considerar aumentar su inversión inicial.")

    # Caso 5: Perfil no válido (riesgo fuera de rango)
    else:
        print("\nIA Robo-Advisor: El nivel de riesgo ingresado no es válido (debe ser del 1 al 10).")

    # Preguntar si desea ejecutar el sistema nuevamente
    ejecutar_sistema = input("\n¿Desea realizar otra evaluación de portafolio? (si/no): ").lower()

print("\n--- Gracias por usar el Simulador de Robo-Advisor IA. ¡Hasta pronto! ---")
```
